## **Many Glacier SNOTEL Station Dataset EDA**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('GNP_SNOTEL_dataset_1986_2026.txt').set_index('Date')
dataset.head()
```

## Flattop Mountain Dataset

```python
flattop_data = dataset.iloc[:, :6]
flattop_data.head()
```

```python
flattop_temp_data = flattop_data.iloc[:, :1].copy()
flattop_temp_data = flattop_temp_data.rename(columns={'Flattop Mtn. (482) Air Temperature Average (degF)': 'average_air_temp_F'})
flattop_temp_data.index = pd.to_datetime(flattop_temp_data.index)
flattop_temp_data
```

```python
print(f'Total Missing Values: {flattop_temp_data.isna().sum()} \n')
print(flattop_temp_data.info())
```

```python
flattop_temp_data = flattop_temp_data.fillna(flattop_temp_data.groupby(flattop_temp_data.index.hour).transform('mean')).copy()
print(f'Total Missing Values After Transform: {flattop_temp_data.isna().sum()}')
```

```python
flattop_temp_data['average_air_temp_C'] = (flattop_temp_data['average_air_temp_F'] - 32) * 5/9
flattop_temp_data
```

```python
flattop_celcius_time_series = flattop_temp_data['average_air_temp_C']
flattop_celcius_time_series
```

```python
flattop_celcius_time_series.to_csv('flattop_average_temps_celcius.csv', index=True)
```

## Many Glacier Dataset

```python
many_glacier_data = dataset.iloc[:, 6:]
many_glacier_data.head()
```

```python
many_glacier_temp_data = many_glacier_data.iloc[:, :1].copy()
many_glacier_temp_data = many_glacier_temp_data.rename(columns={'Many Glacier (613) Air Temperature Average (degF)':'average_air_temp_F'})
many_glacier_temp_data.index = pd.to_datetime(many_glacier_temp_data.index)
many_glacier_temp_data
```

```python
print(f'Total Missing Values: {many_glacier_temp_data.isna().sum()} \n')
print(many_glacier_temp_data.info())
```

```python
many_glacier_temp_data = many_glacier_temp_data.fillna(
    many_glacier_temp_data.groupby(many_glacier_temp_data.index.hour).transform('mean')
).copy()

print(f'Total Missing Values After Transform: {many_glacier_temp_data.isna().sum()}')
```

```python
many_glacier_temp_data['average_air_temp_C'] = (many_glacier_temp_data['average_air_temp_F'] - 32) * 5/9
many_glacier_temp_data
```

```python
many_glacier_celcius_time_series = many_glacier_temp_data['average_air_temp_C']
many_glacier_celcius_time_series
```

```python
many_glacier_celcius_time_series.to_csv('many_glacier_average_temps_celcius.csv', index=True)
```
